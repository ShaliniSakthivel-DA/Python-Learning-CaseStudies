# List

```python
# Creating List

Fruits = ['Apple','Banana','Cherry','Apple']
Fruits[1]

# Modifing List

Fruits[1] = 'Orange'
print(Fruits)

# Add list at end(Append)

Fruits.append('Mango')
print(Fruits)

# Add multiple items (Extend)

Fruits.extend(['Papaya','Mango'])
print(Fruits)

# Add item at specific position

Fruits.insert(2,'Mango')
print(Fruits)

# remove()

Fruits.remove('Mango')
print(Fruits)

# pop()
Fruits.pop()
print(Fruits)

# pop() Revome index item

Fruits.pop(1)
print(Fruits)


```

```python
# index() Find index value

Fruits = ['Apple','Banana','Cherry','Apple']
Fruits.index('Banana')

# Count()
Fruits.count('Apple')
print(Fruits)
```

```python
# sort()

Fruits = ['Mango','Apple', 'Apple', 'Banana', 'Cherry']
Fruits.sort()
print(Fruits)

# Sorted - Create in new variable

Fruits = ['Mango','Apple', 'Apple', 'Banana', 'Cherry']
new_Fruits = sorted(Fruits)
print(new_Fruits)

# reverse

Fruits.sort(reverse= True)
print(Fruits)

# key

Fruits.sort( key= len)
print(Fruits)

```

```python
# Clear()

Fruits.clear()
Fruits
```

```python
# copy()

Fruits = ['Apple','Banana','Cherry','Apple']
Fruits_1 = Fruits.copy()
print(Fruits_1)
```

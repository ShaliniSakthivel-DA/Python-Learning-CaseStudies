```python
# Creating Dictionary
data = {
    'ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank'],
    'Age': [24, 30, 28, 35, 22, 40],
    'City': ['New York', 'London', 'Paris', 'Tokyo', 'Berlin', 'Sydney'],
    'Salary': [50000, 60000, 55000, 70000, 48000, 75000],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'Marketing', 'Finance']
}

df = pd.DataFrame(data)
df
```

```python
#Accessing Columns in Pandas DataFrame

df["Name"]  # Single Element

df[['ID','Name']] # Multiple Elements
```

```python
#Accessing Rows in Pandas DataFrame

df[0:3]
df[2:5]
```

```python
# loc[]  -> Label Based Indexing
# Syntax: df.loc[row_label, column_name]

print("\n---------Single loc------------")

print(df.loc[2,'Name'])
print(df.loc[4,'Department']) # indexing
print(df.loc[3:3])
print(df.loc[0:3]) # Single Row Single coulmn
print("\n---------Multiple loc------------")
print(df.loc[2:3,['Name','Age','City']]) #Multiple Column # slicing
```

```python
# iloc[] -> Integer Position Based Indexing
# Syntax: df.iloc[row_index, column_index]


print(df.iloc[3])

df.iloc[0,3] # ',' give location based on integer(Row, Column)

print(df.iloc[0:3])
```

```python
print(df.loc[0:2]) # loc include end lable
print(df.iloc[0:2]) # loc excludes end lable like slicin
```

```python
print(df.loc[0,'Name'])
print(df.iloc[0,1])
```

```python
# Filtering Data

df['Age'] <= 30

df_age_sal = df[(df['Age'] <=30) & (df['Salary'] <=50000) ]
df_age_sal
```

```python
# Sorting Data
df.sort_values("Age")
```

```python
# Adding Column
df.loc[6] =[107,'George',35,'Paris',55000,"IT"]
df
```

```python
df.loc[len(df)] =[108,'John',25,'Paris',59000,"IT"]
df
```

```python
# Updating Values
df.loc[0,'Salary'] = 40000

df.loc[df['Name']== 'Frank',"Age"] =20
df

df.loc[df['Name']== 'John','Department'] = "HR"
df
```

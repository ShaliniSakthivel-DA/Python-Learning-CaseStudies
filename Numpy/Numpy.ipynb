```python
# STUDENT PERFORMANCE ANALYSIS USING NUMPY

# Import NumPy library
import numpy as np

# Create Student Marks Dataset
# Rows -> Students
# Columns -> Subjects

marks = np.array([
    [78, 85, 90, 88, 76],   # Student A
    [65, 70, 60, 72, 68],   # Student B
    [90, 92, 88, 95, 91],   # Student C
    [45, 50, 55, 48, 52],   # Student D
    [80, 75, 85, 82, 79]    # Student E
])

# Store Student Names
students = np.array(["A", "B", "C", "D", "E"])

# Display Original Dataset
print("========== STUDENT MARKS ==========")
print(marks)

# Calculate Total Marks of Each Student
# axis = 1 -> Row-wise calculation

total_marks = np.sum(marks, axis=1)

print("\n========== TOTAL MARKS ==========")
print(total_marks)

# Calculate Average Marks of Each Student

average_marks = np.mean(marks, axis=1)

print("\n========== AVERAGE MARKS ==========")
print(average_marks)

# Find Highest Mark of Each Student

highest_marks = np.max(marks, axis=1)

print("\n========== HIGHEST MARKS ==========")
print(highest_marks)

# Find Lowest Mark of Each Student

lowest_marks = np.min(marks, axis=1)

print("\n========== LOWEST MARKS ==========")
print(lowest_marks)

# Calculate Percentage
# Total Marks = 500

percentage = (total_marks / 500) * 100

print("\n========== PERCENTAGE ==========")
print(percentage)

# Find Overall Average Total Marks

overall_average = np.mean(total_marks)

print("\n========== OVERALL AVERAGE ==========")
print(overall_average)

# Find Students Scoring Above Overall Average

above_average_students = students[total_marks > overall_average]

print("\n========== ABOVE AVERAGE STUDENTS ==========")
print(above_average_students)

# Find Failed Students
# If any subject mark is less than 50

failed_students = students[np.any(marks < 50, axis=1)]

print("\n========== FAILED STUDENTS ==========")
print(failed_students)

# Calculate Subject-wise Average
# axis = 0 -> Column-wise calculation

subject_average = np.mean(marks, axis=0)

print("\n========== SUBJECT-WISE AVERAGE ==========")
print(subject_average)

# Display Student-wise Report

print("\n========== STUDENT REPORT ==========")

for i in range(len(students)):

    print("\nStudent Name :", students[i])

    print("Marks :", marks[i])

    print("Total Marks :", total_marks[i])

    print("Average Marks :", average_marks[i])

    print("Highest Mark :", highest_marks[i])

    print("Lowest Mark :", lowest_marks[i])

    print("Percentage :", percentage[i])

# Rank Students Based on Total Marks
# argsort() sorts in ascending order
# [::-1] converts to descending order

ranking = np.argsort(total_marks)[::-1]

print("\n========== STUDENT RANKING ==========")

for rank, index in enumerate(ranking, start=1):

    print(f"Rank {rank} : Student {students[index]} "
          f"-> Total Marks = {total_marks[index]}")
```

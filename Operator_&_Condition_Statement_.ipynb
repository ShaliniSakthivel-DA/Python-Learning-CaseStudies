```python
#  Arithimatic Operator

a = 10
b = 20

print(f"\n Addition:  {a+b}")
print(f"subtraction: {a-b}")
print(f"multiplication {a} * {b} : {a*b}")
```

```python
# Comparison Operator

a = 10
b = 20

print(f"\n {a} Greater than {b}: {a>b}")
print(f"\n {a} Less than {b}: {a<b}")
print(f'\n {a} not equal to {b}: {a==b}')
```

```python
# Conditional operator

x = True
y = False

print(x and y)
print(x or y)
```

```python
# Membership Operator in , not in

Text = "I Love Python"

print("Python" in Text)
print("Love" not in Text)
print("Love" in Text)
```

```python
# Identity Operator is , isnot

a = [1,2,3]
b = [2,3,4]

c = a

print (f"{a} and {b} in same memory location: {a is b}")
print (f"{a} and  {b} is not in a same memory location: {a is not b}")
print (f"{a} and {c} is in same memory location: {a is c}")
```

```python
# Assignment Operator

a = 5

a += 5
a -= 1
a *= 2
a >>= 2
a <<= 2

print (a)
```

```python
# Conditional Statements

age = int(input("Enter your Age: "))

if age >= 18:
  print("Your Eligible to Vote")

print("Age Check")
```

```python
# If Else

age = int(input("Enter your Age: "))

if age >= 18:
  print("Your Eligible to Vote")
else:
   print("Not Eligible to Vote")


```

```python
# If else

num = int(input("Enter a number: "))

if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")

```

```python
# else if

num = int(input("Enter a number: "))

if num > 0:
    print("Positive Number")
elif num < 0:
    print("Negative Number")
else:
    print("Zero")
```

```python
# if else

ch = input("Enter a character: ")

if ch in "aeiouAEIOU":
    print("Vowel")
else:
    print("Not a Vowel")
```

```python
# else if

mark = int(input("Enter the mark:"))

if mark >= 90:
  print("Grade A")
elif mark >= 70:
  print("Grade B")
elif mark >= 50:
    print("Grade C")
elif mark >= 40:
    print("Grade D")
else:
    print("Fail")
```

```python
# Vote Eligibility and Citizenship Check using Nested If

age = int(input("Enter your age: "))
citizen = input("Are you a citizen of India? (yes/no): ")

if citizen == "yes":
    if age >= 18:
        print("Eligible to vote")
    else:
        print("Not eligible to vote due to age")
else:
    print("Not eligible because you are not a citizen of India")
```

```python
# Bank Loan Eligibility System

customer_name = input("Enter customer name: ")
age = int(input("Enter age: "))
income = int(input("Enter income: "))
credit_score = int(input("Enter credit score: "))
existing_loan = input("Do you have existing loan? (Yes/No): ")

print("Customer Name:", customer_name)
print("Age:", age)
print("Income:", income)
print("Credit Score:", credit_score)
print("Existing Loan:", existing_loan)

if age >= 21 and income >= 30000 and credit_score >= 700 and existing_loan == "No":
    print("Loan Status: Approved")
elif credit_score >= 600 and income >= 40000:
    print("Loan Status: Conditionally Approved")
else:
    print("Loan Status: Rejected")
```

```python
# Scholarship Eligibility Checker

student_name = input("Enter student name: ")
marks = int(input("Enter marks: "))
attendance = int(input("Enter attendance percentage: "))
family_income = int(input("Enter family income: "))

print("Student Name:", student_name)
print("Marks:", marks)
print("Attendance:", attendance)
print("Family Income:", family_income)

if marks >= 90 and attendance >= 90:
    print("Scholarship: Full Scholarship")
elif marks >= 80 or attendance >= 85:
    print("Scholarship: Half Scholarship")
elif marks >= 70 and family_income < 30000:
    print("Scholarship: Need-Based Scholarship")
else:
    print("Scholarship: Not Eligible")
```

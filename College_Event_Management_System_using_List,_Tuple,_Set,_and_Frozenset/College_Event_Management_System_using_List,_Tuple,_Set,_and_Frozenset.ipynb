```python
# College Event Management System

# 1. Participants List Operations

participants = ["Arun", "Divya", "Kiran"]

# Add participant

participants.append("Ram")

# Insert participant

participants.insert(1,"Shena")

# Remove participant

participants.remove("Ram")

# Display participants

print("Participants List: ",participants)

# 2. Tuple Operations

events = ("Dance", "Coding", "Singing", "Photography")
print("\nFrist Events:", events[0])
print("Last Event:", events[3])
print("Total Number of Events:", len(events))

# 3. Set Operations
departments = ["CSE", "IT", "ECE", "CSE", "IT"]

# Convert list to set to remove duplicates
dept_set = set(departments)

# Add department
dept_set.add("MECH")

# Remove department
dept_set.remove("ECE")

print("\nDepartments Set:")
print(dept_set)

# 4. Frozenset Operations
winners = [("Arun", "Coding"), ("Divya", "Dance"), ("Sneha", "Singing")]

# Convert list into frozenset
winner_frozenset = frozenset(winners)

print("\nWinners Frozenset:")
print(winner_frozenset)

```

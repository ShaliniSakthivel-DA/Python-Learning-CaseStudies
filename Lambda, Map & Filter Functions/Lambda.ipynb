```python
# Sqaure

x = [10, 10, 29, 3, 17, 4]
Square = list(map(lambda a: a * a, x))
print(f"Squared Values: {Square}")
```

```python
x = [3.12234, 14.12233, 5.21111, 3.02333]

num = list(map(lambda a: round(a,1), x))

print(f"Rounded List: {num}")
```

```python
numbers=[1.0459,-2.897,9.87655,9.76798,1.2]
rounded_list=list(map(lambda x : round(x),numbers))
print(f"Original Numbers List : {numbers}\n\nRounded List : {rounded_list}")
```

```python
numbers = [10, 5, 4, 5, 1, 3, 7]

even = list(filter(lambda x: x % 2 == 0, numbers))
print(f"Even Numbers : {even}")

odd = list(filter(lambda x: x % 2 != 0, numbers))
print(f"Odd Numbers : {odd}")
```

```python
letters = ['a', 'b', 'e', 'f', 'i', 'o', 'u', 'z']

# Filter vowels
vowels = list(filter(lambda x: x.lower() in 'aeiou', letters))
print(f"Vowels : {vowels}")

# Filter consonants
consonants = list(filter(lambda x: x.lower() not in 'aeiou', letters))
print(f"Consonants : {consonants}")
```

```python
# Employee Salary Analysis

# Create salary list
salary = [25000, 45000, 55000, 70000, 48000, 90000]

# Display original salary list
print("Original Salary List:", salary)

# Built-in Functions
max_salary = max(salary)
min_salary = min(salary)
total_salary = sum(salary)
employee_count = len(salary)

# Increase each salary by 10% using lambda and map()
updated_salary = list(map(lambda x: x + (x * 0.10), salary))

# Filter salaries greater than 50000 using lambda and filter()
high_salary = list(filter(lambda x: x > 50000, updated_salary))

# Display Results
print("\nUpdated Salary List:", updated_salary)
print("High Salary List (>50000):", high_salary)

print("\nMaximum Salary:", max_salary)
print("Minimum Salary:", min_salary)
print("Total Salary:", total_salary)
print("Number of Employees:", employee_count)
```

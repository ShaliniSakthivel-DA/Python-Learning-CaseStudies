```python
# While Loop

i = 1

while i <= 20:
  print(i)
  i += 1
print(f"Finally Terminated Number: {i}")
```

```python
# Print even numbers from 1 to 100
num = 1
while num <= 100:
  if num % 2 == 0:
    print(num)
  num += 1
```

```python
# Print odd numbers from 1 to 100

num = 1
while num <= 100:
  if num % 2 != 0:
    print(num)
  num += 1
```

```python
students = ['Shalini', 'Arun', 'Agaran']
i = 0

while i <= len(students):
  print(students[i])
  i += 1
```

```python
password = "python123"

user_password = input('Enter the Password: ')
while user_password != password:
    print("Incorrect password... try again")
    user_password = input("Enter the password: ")

else:
    print("Access granted")
    print("Welcome Shalini")
```

```python
# Case Study – ATM PIN Verification System

correct_pin = "1234"
while True:
    user_pin = input("Enter ATM PIN: ")
    if user_pin == correct_pin:
        print("Login Successful")
        break

    else:
        print("Incorrect PIN... Try Again")
```

```python
password = "python123"
attempts = 0
max_attempts = 3

while attempts < max_attempts:
      user_input = input("Enter password: ")

      if user_input == password:
        print("Access granted!")
        break
      else:
        attempts += 1
        print(f"Incorrect password, try again. ({max_attempts - attempts} attempts left)")
if attempts == max_attempts:
    print("Maximum attempts reached. Access denied!")
```

```python
# Printing odd numbers from 1 to 100 using continue

i = 0

while i <= 100:

    i += 1

    # Skip even numbers
    if i % 2 == 0:
        continue

    print(i)
```

```python
# Print students who scored below 50 using continue

students = ["Alice", "Bob", "Charlie", "David", "Eva"]
marks = [45, 78, 32, 66, 49]

i = 0

while i < len(students):

    if marks[i] >= 50:
        i += 1
        continue

    print(f"{students[i]} scored {marks[i]} (Below 50)")

    i += 1

i = 0
```

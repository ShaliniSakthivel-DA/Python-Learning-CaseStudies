# Types of User Defined Functions in Python

```python
# Function without Arguments and without Return Value`

def greet():
  print("Hello Welcome")
greet()
```

```python
#Function with Arguments and without Return Value

def add(a,b): #parameters
  c = a + b
  print("Sum of two",c)


add(10,20) # Arguments
```

```python
#Function without Arguments and with Return Value
def greets():
  return "Hello welcome to Python session"

text = greets()
print(text)
```

```python
#Function with Arguments and with Return Value
def add(a,b): #parameters
  c = a + b
  return c

addition = add(10,20) # Arguments

print(f"Sum of 2 numbers {addition}")
```

```python
# Funtion  with parameters`

def greeting(name = "Learners"):
  print(f"Hello {name} Welcome to Python Session")

greeting("Shalini")
greeting()
```

```python
# Function with multiple returns

def art(a,b): #parameters
  s = a + b
  d = a - b
  return s, d

sum, difference= art(10,20) # Arguments

print("sume of 2 numbers",sum)
print("Diference of 2 numbers",difference)
```

Case Study – Arithmetic Calculator using User Defined Function
Instructions

Create a function called arth_cal()

Pass two parameters to the function

Inside the function:Perform addition, subtraction, multiplication

Return all three values using multiple return statements

Take two numbers from the user using input()

Store returned values in separate variables

Display:Sum, Difference, Multiplication



```python
# Arithmetic Calculation using User Input

def arth_cal(num1, num2):
    addition = num1 + num2
    subtraction = num1 - num2
    multiplication = num1 * num2

    return addition, subtraction, multiplication

# Get input from user
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

# Store returned values
sum_result, diff_result, mul_result = arth_cal(num1, num2)

# Display results
print("Sum =", sum_result)
print("Difference =", diff_result)
print("Multiplication =", mul_result)
```

Case Study – Student Marks Analysis using Function and Loop
Instructions

Create a list called marks

Create a function called marks_analysis() and pass the marks list to the function

Inside the function:

Use a for loop to display all marks

Find total marks

Find average marks

Return total and average marks, call the function, and display: Student Marks, Total Marks, Average Marks


```python
# Marks List
marks = [80, 75, 90, 85, 70]

# Function Creation
def marks_analysis(data):

    print("Student Marks :")

    # for loop
    for i in data:
        print(i)

    # Operations
    total = sum(data)
    average = total / len(data)

    # Return Values
    return total, average


# Function Call
total_marks, average_marks = marks_analysis(marks)

# Output
print("\nTotal Marks   :", total_marks)
print("Average Marks :", average_marks)
```

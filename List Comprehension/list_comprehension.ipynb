```python
# Even number 1 to 100

num = [i for i in range(1,101) if i % 2 == 0]
print(num)
```

```python
# Odd  number 1 to 100

num = [i for i in range(1,101) if i % 2 != 0]
print(num)
```

```python
# String Analysis using List Comprehension

text = "data analytics"

# Convert to uppercase
upper_text = [x.upper() for x in text]

# Extract vowels
vowels = [x for x in text if x in "aeiou"]

# Extract consonants
consonants = [x for x in text if x.isalpha() and x not in "aeiou"]

# Remove spaces
remove_spaces = [x for x in text if x != " "]

print("Uppercase Letters :", upper_text)
print("Vowels            :", vowels)
print("Consonants        :", consonants)
print("Without Spaces    :", remove_spaces)
```

```python
# Original salary list
salary = [25000, 45000, 50000, 60000, 75000]

# Increase each salary by 10%
updated_salary = [s + (s * 0.10) for s in salary]

# Display only salaries greater than 50000
filtered_salary = [s for s in updated_salary if s > 50000]

# Print results
print("Original Salary List:", salary)
print("Updated Salary List:", updated_salary)
print("Filtered Salary List:", filtered_salary)
```

```python
# Filter the positive numbers
x = [-1, -4, 5, 7, 8, -4, -7, 6]

positive_numbers = [i for i in x if i > 0]

print("Positive Numbers:", positive_numbers)


# Filter sales greater than 1000 and return the sum
total_sales = [2000, 300, 200, 500, 5000]

sales_greater_than_1000 = [sale for sale in total_sales if sale > 1000]

total = sum(sales_greater_than_1000)

print("Sales Greater Than 1000:", sales_greater_than_1000)
print("Total Sum:", total)
```

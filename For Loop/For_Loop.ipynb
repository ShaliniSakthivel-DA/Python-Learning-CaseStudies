```python
Text = input("Enter a String: ").lower()

v = c = d = sc = 0

for letter in Text:

  if letter in "aeiou":
    v += 1
  elif letter.isalpha():
    c +=1
  elif letter.isdigit():
    d +=1
  else:
    sc += 1

print(v,c,d,sc)
```

```python
#Case Study – Restricted Word Detection System
#Instructions:
#Create a list of words containing normal and restricted words.
#Use a for loop to read words one by one.
#Check whether the current word is a restricted word using if condition.
#If the restricted word is found: • Display "Restricted word detected"
#• Stop the loop using break
#Otherwise display all safe words using print()#


words = ["Hello", "Python", "Admin", "Blocked", "Welcome"]
restricted_word = "Blocked"

for word in words:
  if word == restricted_word:
    print("Restricted word detected")
    Break
    print(word)
```

```python

words = ["Hello", "Python", "Admin", "Blocked", "Welcome", "Hacker", "Wrong", "Blocked"]

restricted_count = 0

for word in words:

    if word in ["Blocked", "Wrong"]:
        print("Restricted word detected:", word)

        restricted_count += 1


    print("Safe Words:", word)

print("Count of restricted words:", restricted_count)
```

```python
# Case Study – Student Performance Management System

# Create lists
students = ["Shalini", "Arun", "Agaran", "Priya", "Rahul"]
marks = [85, 78, 92, 88, 75]

# Using range()

for i in range(len(students)):
  print("Student:", students[i], "| Marks:", marks[i])

# Using enumerate()

for index, student in enumerate(students):
  print("Index:", index, "| Student Name:", student)

# Using zip()

for student, mark in zip(students, marks):
    print(student, "scored", mark)
```

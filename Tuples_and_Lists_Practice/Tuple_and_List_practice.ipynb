```python
# packing
student = ("Shalini", 26, "Python")
print(student)

```

```python
# Unpacking
name, age, course = student

print(name)
print(age)
print(course)
```

```python
name, age, course = student


student = ("Shalini", 26, "Python")
print(student)

# modifing the age
age = 26 + 1


student = (name, age, course )
print(student)
```

```python
# Packing

order = ("order01", "mobile", 1, 50000)

order_id, product_name, quanity, price = order
print (order)

# Unpacking

order_id, product_name, quanity, price = order
print(order_id)
print(product_name)
print(quanity)

# Modifing

quanity = 1 + 1

order = (order_id, product_name, quanity, price )
print(order)
```
